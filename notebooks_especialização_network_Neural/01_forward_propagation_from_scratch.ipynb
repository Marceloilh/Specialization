Estudo de Rede Neural do 0 ao avançado



```python
import numpy as np
```

```python


class Neural_Network(object):
    def __init__(self):
        # Definição dos parâmetros da rede
        self.inputLayerSize = 2
        self.outputLayerSize = 1
        self.hiddenLayerSize = 3

        # Matrizes de Pesos Iniciais
        self.W1 = np.random.randn(self.inputLayerSize, self.hiddenLayerSize)
        self.W2 = np.random.randn(self.hiddenLayerSize, self.outputLayerSize)

    def forward(self, X):
        # Propagação dos dados
        self.z2 = np.dot(X, self.W1)
        self.a2 = self.sigmoid(self.z2)
        self.z3 = np.dot(self.a2, self.W2)
        yHat = self.sigmoid(self.z3)
        return yHat

    def sigmoid(self, z):
        # Função de Ativação Sigmoid
        return 1 / (1 + np.exp(-z))
```

```python
## 1. Definir os dados de Entrada (X) e a Saída Esperada (y)
X = np.array([
    [3, 5],
    [5, 1],
    [10, 2]
], dtype=float)

y = np.array([
    [0.75],
    [0.82],
    [0.93]
], dtype=float)
```

```python
# Normalização simples (dividir pelos máximos para ficar entre 0 e 1, igual à aula)
X = X / np.amax(X, axis=0)
```

```python
# 3. Instanciar a Rede e Calcular a Previsão
NN = Neural_Network()
yHat = NN.forward(X)
```

```python
# Primeiro foi treinado para um peso de w1 para 1000
# Segundo foi treinado para dois pesos de w1 para 1000 x 1000


import numpy as np
import time 

weightsToTry = np.linspace(-10, 10, 1000)
costs = np.zeros((1000)) # para dois valores 34s adicione 1000

startTime = time.perf_counter()

for i in range(1000):
    #for j in range(1000):
        NN.W1[0, 0] = weightsToTry[i]
        #NN.W1[0, 1] = weightsToTry[j]
        yHat = NN.forward(X)
        # ✅ Alterado 'sum' para 'np.sum'
        costs[i] = 0.5 * np.sum((y - yHat)**2)

endTime = time.perf_counter()

print(f"Tempo total: {endTime - startTime:.4f} segundos")
```

```python
timeElapsed = endTime-startTime
timeElapsed
```

```python
import matplotlib.pyplot as plt

plt.plot(weightsToTry, costs)
plt.grid(True)
plt.ylabel('Custo (Erro)')
plt.xlabel('Peso W1[0,0]')
plt.title('Função de Custo vs. Peso')
plt.show()
```
