<h3> Estruturas de controle e repetição </h3>

```python

for i in range(5):
    print(i)
```

```python
s1 =['P','y','t','h','o','n']
s3 = ['c','a','r','r','o','s']
```

```python
for i , letras in enumerate(s1):
    print(i, letras)
```

```python
for a, letras in zip(s1, s3):
    print(a, letras)
```

```python
i = 1

while i < 6 :
    print(i)
    i += 1
```

```python
for i in range(10,25,3):
    print(i)
```

```python
nota = int(input("Entre a nota (0-100): "))

if nota < 60 :
    print("F")
elif nota < 70:
    print("D")
elif nota < 80:
    print("C")
elif nota < 90:
    print("B")
elif nota <=100:
    print("A")
else:
    print("Nota invalida.")
```

```python
def media(lista):
    return sum(lista)/len(lista)

media([25,50,100])
```

```python
def imprimir_ate(lista, n=5):
    print(lista[:n])
imprimir_ate(list(range(1, 100)))
```

<h3> Classes </h3>

```python
class Gato:
    name = "Garfield"
    old = "10"
    color = "yellow"

meuGato = Gato()

dados_gato = meuGato.name, meuGato.old, meuGato.color
print(dados_gato)
```

```python
class Cachorro:
    def __init__(self, name, raca):
        self.name = name
        self.raca = raca
    def latir(self):
        print(f'"Woof!" - exclamou {self.name}.')

meuCachorro = Cachorro("Marley", "Labrador")
```

```python
meuCachorro.latir()
```
