<h3>Exercício 1. Peça ao usuário para digitar uma string e imprima se esta string
é um palíndromo ou não. (Palíndromo é uma palavra que é lida da esquerda para
a direita, conforme o sentido habitual da leitura, ou da direita para esquerda).</h3
                                                                                    >

```python
import re
import unicodedata

def limpar_texto(texto):
    # 1. Remove acentos, til e cedilha
    texto_decomposto = unicodedata.normalize('NFD', texto)
    texto_limpo = ''.join(c for c in texto_decomposto if unicodedata.category(c) != 'Mn')
    
    # 2. Divide no '@' ou no ponto '.'
    palavras_divididas = re.split(r'[@.]', texto_limpo)
    
    return palavras_divididas

def checar_palindromo(palavra):
    # Padroniza para minúsculas e remove espaços internos
    texto = palavra.lower().replace(" ", "")
    
    # Retorna True se for igual ao seu inverso
    return texto == texto[::-1]

# --- Execução do Programa ---
palavra_digitada = input("Digite sua palavra: ")

# Limpa e divide o texto digitado
palavras_limpas = limpar_texto(palavra_digitada)

# Processa cada item resultante da divisão
for palavra in palavras_limpas:
    tamanho = len(palavra)
    eh_palindromo = checar_palindromo(palavra)
    
    print(f"\nItem: '{palavra}'")
    print(f"Tamanho: {tamanho}")
    print(f"É palíndromo? {eh_palindromo}")
   









```

<h3>Exercício 2. Escreva um programa que pergunte ao usuário quantos números
de Fibonacci serão gerados. Aproveite esta oportunidade para pensar sobre como
você pode usar funções. Certifique-se de pedir ao usuário para inserir o número
de números na sequencia para gerar. (Dica: A sequencia Fibonacci é uma sequen
cia de números onde o próximo número na sequencia é a soma dos dois números
anteriores na sequencia. é assim: 1, 1, 2, 3, 5, 8, 13,…).</h3>


```python
# 1. Módulo / Regra de Negócio (isolado, não sabe quem é o usuário)
def fibonacci(numero: int) -> list:
    anterior = 0
    atual = 1
    sequencia = []
    
    for _ in range(numero):
        proximo = anterior + atual
        sequencia.append(proximo)
        anterior = atual
        atual = proximo
        
    return sequencia

# 2. Programa Principal / Interface (onde o usuário interage)
numero_user = int(input('Digite a quantidade de números desejada: '))

# Aqui acontece a "ponte": o valor de 'numero_user' é passado para 'numero'
resultado = fibonacci(numero_user)

print(resultado)
```

<h3>Exercício 3. Forca é um jogo muito popular que todos provavelmente conhecem e já jogaram quando crianças. Neste jogo, um jogador define uma palavra</h3>


```python
import random
import unicodedata
import matplotlib.pyplot as plt

# =====================================================================
# 1. VISUALIZAÇÃO GRÁFICA (DESENHO DA FORCA VIA MATPLOTLIB)
# =====================================================================

def desenhar_forca_grafica(erros: int):
    """
    Objetivo: Renderizar visualmente a estrutura da forca e as partes do boneco
              no Jupyter Notebook usando coordenadas do Matplotlib.
    
    Recebe:
        erros (int): Quantidade de erros acumulados no jogo (de 0 a 6).
    """
    # Cria a figura gráfica com proporção (largura=3, altura=4)
    fig, ax = plt.subplots(figsize=(3, 4))
    
    # --- DESENHO DA ESTRUTURA DE MADEIRA ---
    # ax.plot([x1, x2], [y1, y2], 'k-', lw=3) -> 'k-' cria linha preta, lw=espessura
    ax.plot([1, 5], [1, 1], 'k-', lw=3)  # Base da forca (linha horizontal inferior)
    ax.plot([2, 2], [1, 9], 'k-', lw=3)  # Mastro/Poste vertical
    ax.plot([2, 4], [9, 9], 'k-', lw=3)  # Viga superior horizontal
    ax.plot([4, 4], [9, 8], 'k-', lw=2)  # Corda suspensa

    # --- DESENHO PROGRESSIVO DO BONECO (Conforme o número de erros) ---
    if erros >= 1: 
        # Erro 1: Cabeça (círculo com centro na coordenada (4, 7.3) e raio 0.7)
        cabeca = plt.Circle((4, 7.3), 0.7, color='black', fill=False, lw=2)
        ax.add_patch(cabeca)
        
    if erros >= 2: 
        # Erro 2: Tronco/Corpo (linha vertical do pescoço ao quadril)
        ax.plot([4, 4], [6.6, 4.5], 'k-', lw=2)
        
    if erros >= 3: 
        # Erro 3: Braço Esquerdo
        ax.plot([4, 3.2], [6, 5], 'k-', lw=2)
        
    if erros >= 4: 
        # Erro 4: Braço Direito
        ax.plot([4, 4.8], [6, 5], 'k-', lw=2)
        
    if erros >= 5: 
        # Erro 5: Perna Esquerda
        ax.plot([4, 3.2], [4.5, 3], 'k-', lw=2)
        
    if erros >= 6: 
        # Erro 6: Perna Direita (Completa o boneco no enforcamento)
        ax.plot([4, 4.8], [4.5, 3], 'k-', lw=2)

    # --- CONFIGURAÇÃO E LIMPEZA DA TELA DO GRÁFICO ---
    ax.set_xlim(0, 6)   # Limite do eixo X
    ax.set_ylim(0, 10)  # Limite do eixo Y
    ax.axis('off')      # Esconde os eixos, números e bordas do gráfico
    
    # Força o Matplotlib a exibir a imagem imediatamente no Notebook
    plt.show()


# =====================================================================
# 2. FUNÇÕES UTILITÁRIAS E REGRAS DE NEGÓCIO
# =====================================================================

def normalizar_texto(texto: str) -> str:
    """
    Objetivo: Remover acentos, cedilhas e converter o texto para maiúsculas.
    Exemplo: 'maçã' -> 'MACA'
    """
    # 1. Decompõe o caractere em letra base + marca do acento (NFD)
    texto_decomposto = unicodedata.normalize('NFD', texto)
    
    # 2. Filtra mantendo apenas os caracteres que NÃO são acentos (categoria 'Mn')
    texto_limpo = ''.join(c for c in texto_decomposto if unicodedata.category(c) != 'Mn')
    
    # 3. Retorna o texto padronizado em caixa alta
    return texto_limpo.upper()


def carregar_palavras() -> list[str]:
    """
    Objetivo: Definir a lista de palavras do jogo e aplicar a limpeza em todas.
    """
    palavras_brutas = ['carro', 'avião', 'dentista', 'semaforo', 'maçã']
    
    # Aplica a função normalizar_texto em cada item da lista via List Comprehension
    return [normalizar_texto(p) for p in palavras_brutas]


def sortear_palavra(lista_palavras: list[str]) -> str:
    """
    Objetivo: Sortear uma palavra aleatória da lista tratada.
    """
    return random.choice(lista_palavras)


def criar_mascara(palavra_secreta: str) -> list[str]:
    """
    Objetivo: Criar a lista de underlines com a quantidade exata de letras da palavra.
    Exemplo: 'MACA' -> ['_', '_', '_', '_']
    """
    return ['_'] * len(palavra_secreta)


def validar_chute(chute: str, letras_chutadas: list[str]) -> bool:
    """
    Objetivo: Garantir que o chute do jogador seja válido antes de processar a jogada.
    """
    chute = chute.upper().strip()

    # Checa se o usuário digitou exatamente 1 caractere e se ele é uma letra do alfabeto
    if len(chute) != 1 or not chute.isalpha():
        print("⚠️ Entrada inválida! Digite apenas UMA letra.")
        return False

    # Checa se a letra já está no histórico de tentativas
    if chute in letras_chutadas:
        print(f"⚠️️ Você já chutou a letra '{chute}'. Tente outra!")
        return False

    return True


def atualizar_mascara(palavra_secreta: str, mascara: list[str], chute: str) -> list[str]:
    """
    Objetivo: Revelar o chute nas posições correspondentes dentro da máscara visual.
    """
    # O enumerate percorre a palavra retornando a posição (índice) e o caractere (letra)
    for indice, letra in enumerate(palavra_secreta):
        if letra == chute:
            mascara[indice] = chute  # Substitui o '_' pela letra revelada
            
    return mascara


# =====================================================================
# 3. FLUXO PRINCIPAL DA PARTIDA (GERENCIADOR DO LOOP)
# =====================================================================

def jogar_forca():
    """
    Objetivo: Controlar o ciclo de vida do jogo (início, rodadas, renderização e fim).
    """
    # --- CONFIGURAÇÃO INICIAL DA PARTIDA ---
    palavras = carregar_palavras()                # Carrega o banco de palavras limpas
    palavra_secreta = sortear_palavra(palavras)  # Sorteia a palavra da rodada
    mascara = criar_mascara(palavra_secreta)     # Gera a máscara ['_', '_', ...]
    
    tentativas = 6                               # Vidas iniciais do jogador
    letras_chutadas = []                         # Histórico de chutes

    print("=" * 35)
    print("      JOGO DA FORCA EM PYTHON     ")
    print("=" * 35)

    # --- CICLO DE VIDA DA PARTIDA (LOOP PRINCIPAL) ---
    while tentativas > 0 and '_' in mascara:
        # 1. Calcula os erros atuais (Total de Vidas - Vidas Restantes)
        erros = 6 - tentativas
        
        # 2. Desenha o boneco na forca graficamente via Matplotlib
        desenhar_forca_grafica(erros)

        # 3. Exibe o estado textual do jogo
        print(f"\nPalavra: {' '.join(mascara)}")
        print(f"Tentativas restantes: {tentativas}")
        print(f"Letras já tentadas: {', '.join(letras_chutadas) if letras_chutadas else 'Nenhuma'}")

        # 4. Recebe a entrada do jogador (No Jupyter, abre a caixa de diálogo no topo)
        chute = input("Digite uma letra: ").upper().strip()

        # 5. Se a entrada for inválida, pula o resto da rodada sem gastar vidas
        if not validar_chute(chute, letras_chutadas):
            continue

        # 6. Registra o chute válido no histórico
        letras_chutadas.append(chute)

        # 7. Processa Acerto ou Erro
        if chute in palavra_secreta:
            print(f"✅ Boa! A letra '{chute}' está na palavra.")
            mascara = atualizar_mascara(palavra_secreta, mascara, chute)
        else:
            tentativas -= 1
            print(f"❌ Que pena! A letra '{chute}' não está na palavra.")

    # --- FINALIZAÇÃO DO JOGO (Fora do while) ---
    # Exibe a figura final da forca (completa se perdeu, ou no estágio atual se venceu)
    desenhar_forca_grafica(6 - tentativas)

    print("\n" + "=" * 35)
    if '_' not in mascara:
        print(f"🎉 PARABÉNS! Você venceu! A palavra era: {palavra_secreta}")
    else:
        print(f"💀 GAME OVER! Você esgotou suas tentativas. A palavra era: {palavra_secreta}")
    print("=" * 35)


# =====================================================================
# 4. EXECUÇÃO
# =====================================================================
# No Jupyter Notebook, executamos diretamente a chamada da função:
jogar_forca()
```

<h4>
    
1. Tratamento e Normalização de Strings (unicodedata e NFD)
   
Um dos maiores desafios ao processar texto em português é a presença de acentos e caracteres especiais (ç, ã, é). Se a palavra secreta for MAÇÃ e o jogador chutar a letra A, o Python sem tratamento entenderia que A é diferente de Ã, penalizando o jogador injustamente.

Para resolver isso, utilizamos a biblioteca nativa unicodedata aplicando o padrão de decomposição Unicode:

O que é o NFD (Normalization Form Canonical Decomposition)?
No padrão Unicode, caracteres acentuados podem ser representados de duas formas:

Composta (NFC): O caractere ã é armazenado como um único bloco de código (U+00E3).

Decomposta (NFD): O caractere ã é separado em dois elementos distintos: a letra base a (U+0061) + a marca do acento til ~ (U+0303).

O Filtro por Categoria (Mn)
Ao aplicar unicodedata.normalize('NFD', texto), separamos as letras de suas marcas de acentuação. Em seguida, o Python classifica as marcas de acento na categoria Unicode Mn (Mark, Nonspacing).

Usando uma List Comprehension, filtramos a string para manter apenas o que não for acento:

Python
texto_limpo = ''.join(c for c in texto_decomposto if unicodedata.category(c) != 'Mn')
Isso transforma 'MAÇÃ' em 'MACA', permitindo comparações diretas, seguras e insensíveis a acentos ou diferenças entre maiúsculas e minúsculas.

2. Ícones no Terminal vs. Renderização Vetorial no Matplotlib
No projeto, utilizamos duas abordagens complementares para a parte visual:

A. Ícones Unicode (Emojis no Terminal)
Símbolos como a caveira 💀, a festa 🎉 e o alerta ⚠️ são caracteres especiais da tabela UTF-8. O terminal ou o ambiente do Jupyter os renderiza diretamente a partir do texto impresso pelo print(). Podem ser usados digitando o emoji diretamente ou via código de escape Unicode (ex: \U0001F480 para 💀).

B. Renderização Gráfica Dinâmica (Matplotlib)
Para desenhar a forca e o boneco no Jupyter Notebook, utilizamos elementos geométricos vetoriais em um plano cartesiano 2D:

Linhas (ax.plot): Usadas para a estrutura de madeira, corda e membros do boneco.

Formas Geométricas (plt.Circle e add_patch): Usado para desenhar a cabeça do boneco.

Lógica Progressiva: A função recebe a variável erros (de 0 a 6) e utiliza condicionais (if erros >= 1, etc.) para desenhar apenas as partes do corpo correspondentes aos erros acumulados na partida.

Limpeza de Interface (ax.axis('off')): Removemos os eixos, marcas e números para transformar um gráfico científico em uma ilustração limpa.

3. Arquitetura do Código: Por que Funções Puras e Modulares?
O programa foi estruturado no paradigma Procedural Modular, seguindo o Princípio da Responsabilidade Única (Single Responsibility Principle).

Benefícios do Estilo Adotado
Pequenas Funções Especializadas: Cada função resolve apenas um problema (validar_chute apenas valida; atualizar_mascara apenas altera a palavra; desenhar_forca_grafica apenas cuida da imagem).

Reutilização do Código: A função normalizar_texto, por exemplo, não tem dependência com o jogo da forca. Ela é um módulo genérico que pode ser copiado e reutilizado em qualquer outro projeto de dados.

Manutenibilidade: Se você quiser alterar a interface visual de Matplotlib para uma biblioteca como Pygame ou Tkinter, precisará modificar apenas a função visual, sem impactar a regra de negócio do jogo.

Por que não utilizamos outros modelos?
Sem Código Monolítico (Script Único): Escrever o jogo inteiro dentro de um único bloco while sem funções gera o chamado "código espaguete", difícil de ler, testar e encontrar erros.

Sem Programação Orientada a Objetos (POO) neste momento: Criar classes como class JogoForca ou class Jogador adicionaria uma camada de abstração desnecessária para o escopo deste projeto, embora seja o passo seguinte ideal para expansões maiores (como adicionar múltiplos jogadores ou partidas em rede).

4. Aplicações Práticas desse Tipo de Código no Mercado
Embora o projeto seja um jogo pedagógico, as técnicas aplicadas são amplamente utilizadas na indústria de tecnologia e engenharia de dados:

Pipelines de Dados e ETL (Data Engineering): A sanitização de strings via unicodedata é uma etapa obrigatória de higienização de dados para padronizar nomes, endereços e cadastros antes da inserção em Data Warehouses.

Mecanismos de Busca (Search Engines): Motores de busca (como Elasticsearch) aplicam normalização NFD para garantir que a pesquisa do usuário por "pao" encontre registros salvos como "Pão".

Desenvolvimento de Dashboards e Visualização de Dados: O uso do Matplotlib para plotar elementos personalizados dinamicamente com base em estados e variáveis é a base para a criação de relatórios automatizados em Data Science.

Validação e Segurança de Entradas: As checagens de integridade (len == 1, isalpha(), strip()) simulam a camada de validação que protege APIs e formulários web contra dados corrompidos ou maliciosos.</h4>

```python
import re

def limpar_e_mapear_palavra(texto_sujo: str, mapa_substituicao: dict) -> list[tuple[str, int]]:
    """
    1. Substitui caracteres leetspeak mapeados.
    2. Remove espaços em branco nas extremidades e converte para maiúsculas.
    3. Retorna a palavra limpa e a lista de tuplas (caractere, indice).
    """
    # 1. Substitui cada caractere mapeado no dicionário
    texto_processado = texto_sujo
    for caractere_sujo, caractere_limpo in mapa_substituicao.items():
        texto_processado = texto_processado.replace(caractere_sujo, caractere_limpo)
    
    # 2. Limpa espaços e converte para maiúsculas
    palavra_limpa = texto_processado.strip().upper()
    
    # 3. Gera a lista de tuplas (caractere, índice) usando enumerate
    mapeamento_indices = [(caractere, indice) for indice, caractere in enumerate(palavra_limpa)]
    
    print(f"Palavra Limpa: {palavra_limpa}")
    return mapeamento_indices

# --- Teste do Desafio 1 ---# Dicionário de mapeamento ORDENADO (dos símbolos maiores para os menores)
mapa = {
    # 1. Combinações de múltiplos caracteres (devem ser substituídas primeiro)
    r'\/': 'V',
    '|_|': 'U',
    r'|V|': 'M',
    '|>': 'P',
    '|<': 'K',
    '(0': 'CO',
    r'|\|': 'N',
    '|2': 'R',
    '|)': 'D',
    '8': 'B',
    '|_': 'L',

    # 2. Caracteres individuais
    '4': 'A',
    '5': 'S',
    '3': 'E',
    '(': 'C',
    '7': 'T',
    '>': 'X',
    '<': 'C',
    'á': 'Á',
    '0': 'O',
    'ú': 'Ú'
}
    
texto_teste = r"  53 \/0(ê |_|54 |_||\/| 73>70 3|\/| |_3375|>34|<, \/0(ê 357á (0|\|\/3|273|\||)0 |_37|245 3|\/| |\|ú|\/|3|205 3 5í|\/|80|_05!  "

resultado = limpar_e_mapear_palavra(texto_teste, mapa)
print("Resultado Mapeado:", resultado)
```
